# STEP 4: Sentiment Analysis – Finding Emotional Causes Behind Drop-Off

#### This step answers the core WHY question:

* "What was the customer feeling before they stopped engaging?"

#### 4️⃣ Why Sentiment Analysis Matters (Business View)

Traditional churn analysis:

* Looks at numbers only

* Misses emotion

This project:

* Combines behavior + emotion

* Detects churn before cancellation

📌 Companies lose customers mostly due to bad experiences, not random inactivity.

#### 4.1 Data We Use

From the dataset:

* feedback_comment

* days_since_last_activity

* churned

#### 4.2 Sentiment Categories (Simple & Effective)

We classify each comment into:

* Positive

* Neutral

* Negative

This is perfect for Business Analyst roles (no overengineering).

#### 4.3 Python: Sentiment Analysis Code (Resume-Ready)

* We’ll use VADER (designed for short customer comments).

```python
import pandas as pd
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk

nltk.download('vader_lexicon')

# Load dataset
df = pd.read_csv("customer_dropoff_analysis_dataset.csv")

# Initialize sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Get sentiment score
df['sentiment_score'] = df['feedback_comment'].apply(
    lambda x: sia.polarity_scores(x)['compound']
)

# Convert score to label
def sentiment_label(score):
    if score >= 0.05:
        return 'Positive'
    elif score <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['sentiment'] = df['sentiment_score'].apply(sentiment_label)

df.head()
```

#### 4.4 Sentiment Distribution (First Insight)

```python
df['sentiment'].value_counts()
```

🧠 Business Insight Example

* “Over 40% of dropped-off customers expressed negative sentiment before disengagement.”

#### 4.5 Sentiment vs Drop-Off Timing (VERY IMPORTANT)

```python
df.groupby('sentiment')['days_since_last_activity'].mean()
```

📌 What This Shows

* Negative sentiment users disengage faster

* Positive users stay active longer

💬 Interview Insight:

* “Emotion precedes disengagement.”

#### 4.6 Sentiment vs Churn Probability

```python
df.groupby('sentiment')['churned'].mean() * 100
```

##### 📊 Example Interpretation

* Negative sentiment → 65% churn rate

* Neutral → 40%

* Positive → 20%

🔥 This is gold for marketing teams.

#### 4.7 Silent Risk Customers (Advanced Insight)

* Customers who sound happy but disappear:

# STEP 5: Customer Drop-Off Personas (Actionable Segmentation)

This step converts raw analysis → clear business actions.

Managers don’t act on rows of data — they act on personas.

#### 🎯 What Is a Drop-Off Persona?

A Drop-Off Persona is a group of customers who:

* Stop engaging for similar reasons

* Show similar behavior patterns

* Need similar retention actions

Instead of saying:
❌ “10,000 customers churned”

You say:
✅ “These 3 types of customers are disengaging — and here’s what to do.”

🧩 Inputs Used to Build Personas

We combine insights from:

* last_activity

* days_since_last_activity

* discount_received & discount_used

* sentiment

* engagement_day

* churned

This is cross-functional analysis (marketing + product + ops).

##### 🧠 Persona 1: Discount-Dependent Drop-Offs
🔍 Characteristics

* Received discounts frequently

* Rarely used discounts

* High inactivity after offers ended

* Medium to high churn probability

📊 Data Signals

* discount_received = 1

* discount_used = 0

* days_since_last_activity ≥ 30

💥 Root Cause

👉 Customers attracted by discounts, not value
👉 Leave once incentives stop

🎯 Business Action

* Reduce blanket discounts

* Offer value-based bundles

* Personalize offers instead of mass promotions

##### 🧠 Persona 2: Experience-Driven Drop-Offs
🔍 Characteristics

* Negative sentiment comments

* Last activity often support ticket / browsing

* Short time between complaint and inactivity

📊 Data Signals

* sentiment = Negative

* Feedback like “Delivery delayed”, “App crashes”

* Rapid engagement drop

💥 Root Cause

👉 Product or service failure
👉 Issue not resolved fast enough

🎯 Business Action

* Priority support routing

* SLA-based follow-ups

* Fix high-frequency complaint themes

##### 🧠 Persona 3: Silent Drop-Offs (Most Dangerous)
🔍 Characteristics

* Positive or neutral feedback

* No complaints

* Long inactivity gap

* Quiet churn

📊 Data Signals

* sentiment = Positive / Neutral

* days_since_last_activity ≥ 40

* No support interactions

💥 Root Cause

👉 Gradual loss of interest
👉 No emotional trigger — just disengagement

🎯 Business Action

* Re-engagement nudges

* Content refresh

* Usage reminders before inactivity threshold

##### 🧠 Persona 4: Timing-Mismatch Users
🔍 Characteristics

* Active mostly on weekends

* Low engagement during weekdays

* Miss marketing campaigns

📊 Data Signals

* engagement_day = Weekend

* Higher drop-off rate

💥 Root Cause

👉 Campaign timing mismatch
👉 Notifications sent at wrong times

🎯 Business Action

* Weekend-focused campaigns

* Time-zone & habit-aware messaging

#### 🧪 Python: Persona Tagging Logic :

```python
def assign_persona(row):
    if row['discount_received'] == 1 and row['discount_used'] == 0 and row['days_since_last_activity'] >= 30:
        return 'Discount-Dependent Drop-Off'
    elif row['sentiment'] == 'Negative' and row['days_since_last_activity'] >= 20:
        return 'Experience-Driven Drop-Off'
    elif row['sentiment'] in ['Positive', 'Neutral'] and row['days_since_last_activity'] >= 40:
        return 'Silent Drop-Off'
    elif row['engagement_day'] == 'Weekend' and row['days_since_last_activity'] >= 30:
        return 'Timing-Mismatch User'
    else:
        return 'Active / Low Risk'

df['dropoff_persona'] = df.apply(assign_persona, axis=1)

df['dropoff_persona'].value_counts()
```
